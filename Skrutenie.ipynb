```python
import numpy as np

# 1. Vstupné parametre (Zadanie)
Mk = 500              # Krútiaci moment [Nm]
L = 1.5               # Dĺžka hriadeľa [m]
d = 0.038             # Navrhnutý priemer hriadeľa (38 mm) [m]

# Materiálové vlastnosti (Oceľ)
medza_klzu_smyk = 120e6
bezpecnost_limit = 2.0
G = 80e9

# Technické obmedzenie na deformáciu
max_uhol_na_meter = 0.25

# ==============================================================================
# 2. VÝPOČTOVÁ ČASŤ
# ==============================================================================
Wp = (np.pi * d**3) / 16
Ip = (np.pi * d**4) / 32

# A. PEVNOSTNÁ KONTROLA
tau_max = Mk / Wp
dovolene_tau = medza_klzu_smyk / bezpecnost_limit
vyhovuje_pevnost = tau_max <= dovolene_tau
vypocitana_bezpecnost = medza_klzu_smyk / tau_max

# B. DEFORMAČNÁ KONTROLA
fi_rad = (Mk * L) / (G * Ip)
fi_deg = np.degrees(fi_rad)
skrutenie_na_meter = fi_deg / L

max_povoleny_uhol = max_uhol_na_meter * L
vyhovuje_deformacia = fi_deg <= max_povoleny_uhol

# ==============================================================================
# 3. VÝSTUPNÝ REPORT
# ==============================================================================
print(f"--- VÝSLEDKY KONTROLY HRIADEĽA (Priemer d = {d*1000:.1f} mm) ---\n")

print("1. PEVNOSŤ:")
print(f"  Maximálne šmykové napätie: {tau_max / 1e6:.2f} MPa")
print(f"  Dovolené šmykové napätie:  {dovolene_tau / 1e6:.2f} MPa")
print(f"  Vypočítaná bezpečnosť:      {vypocitana_bezpecnost:.2f} (Požadovaná: {bezpecnost_limit:.1f})")
if vyhovuje_pevnost:
    print("  => PEVNOSTNÁ PODMIENKA:    VYHOVUJE")
else:
    print("  => PEVNOSTNÁ PODMIENKA:    NEVYHOVUJE! Hriadeľ neunesie toto napätie.")

print("\n2. DEFORMÁCIA:")
print(f"  Celkový uhol skrútenia:    {fi_deg:.3f}° na dĺžke {L} m")
print(f"  Skrútenie na meter dĺžky:  {skrutenie_na_meter:.3f}°/m")
print(f"  Maximálne povolené:        {max_uhol_na_meter:.3f}°/m")
if vyhovuje_deformacia:
    print("  => DEFORMAČNÁ PODMIENKA:   VYHOVUJE")
else:
    print("  => DEFORMAČNÁ PODMIENKA:   NEVYHOVUJE! Hriadeľ sa krúti príliš.")

print("\n" + "="*50)
if vyhovuje_pevnost and vyhovuje_deformacia:
    print("ZÁVER: HRIADEĽ JE SPRÁVNE NAVRHNUTÝ A VYHOVUJE.")
else:
    print("ZÁVER: ROZMERY HRIADEĽA MUSÍTE UPRAVIŤ (Zväčšiť priemer d).")
print("="*50)
```
